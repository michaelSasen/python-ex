TASK 2: DICE GAME

```python
import random as rnd   # imported random as rnd so rnd.randint() can be used

dice = 0
guess = 0

def getDiceGuess():

    guess = 0   # changed guesss -> guess for consistency

    while guess not in ('1','2','3','4'):   # removed spaces after numbers ('1 ' -> '1')
        print('Guess the number on a 4-sided dice! Enter a number between 1 and 4:')
        # removed broken linebreak inside the print statement

        guess = input()

    return int(guess)   # changed from return guesss
                         # converts string input into integer

def rollDice():

    dice = rnd.randint(1, 4)
    # changed random.randint -> rnd.randint because random was imported as rnd
    # changed range from 1-5 to 1-4 because it is a 4-sided dice

    guess = getDiceGuess()
    # added function call to actually get user input

    if dice == guess:
        print('You are good!')
        return
        # added return so the game stops if the player wins

    else:
        print('Sorry! Try to guess again!')

    guess = getDiceGuess()
    # changed input() -> getDiceGuess() to validate input properly

    if dice == guess:
        print('Congrats! You got it!')
        return

    else:
        print('Sorry! Your second chance!')
        # changed "third chance" because this is actually second attempt

        guess = getDiceGuess()

    if dice == guess:
        print('Congrats! You got it!')
        return

    else:
        print('Sorry! Your can still try!')
        # typo/grammar

        guess = getDiceGuess()

    if dice == guess:
        print('Congrats! You got it!')
        return

    else:
        print('Nope. You are really bad at this game.')

# added function call so the program actually starts
rollDice()
```
