```python
# ***********************************************************************
# Sources:
#
# Used for f-formatting strings:
# https://www.w3schools.com/python/python_string_formatting.asp
#
# ***********************************************************************

# Import NumPy library for the matrix operations
import numpy as np

def trace(A):

    # Checks for rank 2 numpy array (matrix)
    if A.ndim != 2:
        print("Input is not a matrix!")
        return

    rows, columns = A.shape

    # Checks if a matrix is a quadratic
    if rows != columns:
        print(f"This is not a quadratic matrix! Shape is {rows}x{columns}")
        return
    print(f"This is a quadratic matrix with the shape {rows}x{columns}")

    # Sums the main diagonal elements for trace
    diagonal_sum = 0

    # Looping through each row and add the diagonal element A[i][i] to the sum
    for i in range(rows):
        diagonal_sum += A[i][i]

    print(f"Trace of the diagonal sum: {diagonal_sum}")

# Quadratic matrix
array1 = np.array([
    [6, 3, 5],
    [1, 9, 2],
    [4, 8, 6]
])

# Not quadratic
array2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Not a matrix (ndim 1)
array3 = np.array(
    [2, 8, 6]
)

print("------ Task 4 ------\n")

print("Array 1: ")
trace(array1) # output 21

print("\n--------------------\n")

print("Array 2: ")
trace(array2) # output not quadratic

print("\n--------------------\n")

print("Array 3: ")
trace(array3) # output not a matrix
```
