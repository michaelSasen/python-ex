```python
# ***********************************************************************
# Sources:
# https://www.codecademy.com/resources/docs/numpy/ndarray/ndim
# https://numpy.org/doc/stable/user/absolute_beginners.html
# https://www.w3schools.com/python/python_string_formatting.asp
# ***********************************************************************

# Import NumPy library for the matrix operations
import numpy as np

def trace(A):

    # Checks for rank 2 numpy array (matrix)
    if A.ndim != 2:
        print("Input is not a 2d matrix!")
        return

    rows, columns = A.shape

    # Checks if a matrix is quadratic
    if rows != columns:
        print(f"Matrix is not quadratic! Shape is {rows}x{columns}")
        return
    print(f"Matrix is quadratic with the shape {rows}x{columns}")

    # Sums the main diagonal elements for trace
    diagonal_sum = 0

    # Looping through each row and add the diagonal element A[i][i] to the sum
    for i in range(rows):
        diagonal_sum += A[i][i]

    print(f"Trace of the diagonal sum: {diagonal_sum}")

# Quadratic matrix
matrix1 = np.array([
    [6, 3, 5],
    [1, 9, 2],
    [4, 8, 6]
])

# Not quadratic
matrix2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Smaller quadratic
matrix3 = np.array([
    [2, 8],
    [6, 3]
])

print("------ Task 4 ------\n")

print("Matrix 1: ")
trace(matrix1) # output 21

print("\n--------------------\n")

print("Matrix 2: ")
trace(matrix2) # output not quadratic

print("\n--------------------\n")

print("Matrix 3: ")
trace(matrix3) # output 5
```
